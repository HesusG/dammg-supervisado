## Parte B. Con Python (5 min)

Ahora comprobamos las cuentas con scikit-learn. Primero convertimos las palabras en 1 y 0: **1 = Adelie** (el positivo).

```python
real = (pinguinos.es == "Adelie").astype(int)               # 1 si de verdad es Adelie
dijo = (pinguinos.el_modelo_dijo == "Adelie").astype(int)   # 1 si el modelo dijo Adelie
```

La tabla de cuatro casillas. Primero va lo que **es**, luego lo que **dijo**. Se lee así: arriba `[VN, FP]`, abajo `[FN, VP]`.

```python
confusion_matrix(real, dijo)   # debe coincidir con su papel
```

```python
accuracy_score(real, dijo)     # exactitud: aciertos entre total
```

```python
precision_score(real, dijo)    # de los que dijo Adelie, cuántos eran
```

```python
recall_score(real, dijo)       # de los Adelie que hay, cuántos encontró
```
