## 1. Los tickets

Tres compras de tortillas: cuántos kilos y cuánto se pagó. La tortillería tiene un letrero de 25 pesos el kilo.

**¿Los tres tickets pagaron lo mismo por kilo?**

*Tip: antes de correr, dividan de cabeza lo pagado entre los kilos.*

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.array([1, 2, 3])      # kilos
y = np.array([25, 51, 74])   # pesos pagados
pd.DataFrame({"x: kilos": x, "y: pagado": y, "pesos por kilo": (y / x).round(2)})
```

## 2. El costo de una recta

Con $b = 0$ la recta es $\hat y = w\,x$, y $w$ es el precio por kilo. Esta celda calcula
$J(w) = \frac{1}{2m}\sum (\hat y - y)^2$ y muestra cada paso, ticket por ticket. Aquí con $w = 25$.

**¿Qué quiere decir que J(25) salga 0.33?**

*Tip: miren la columna error²: ¿qué tickets suman al costo y cuál no?*

```python
def costo(x, y, w):
    error = w * x - y                  # ŷ − y
    return np.mean(error ** 2) / 2     # (1/2m) · Σ error²


def ver_costo(w, x=x, y=y):
    y_hat = w * x
    tabla = pd.DataFrame({"x": x, "y": y, "ŷ = w x": y_hat, "error ŷ − y": y_hat - y, "error²": (y_hat - y) ** 2})
    display(tabla)
    print(f"m = {len(x)}, suma de errores² = {((y_hat - y) ** 2).sum()},  J({w}) = {costo(x, y, w):.2f}")


ver_costo(25)
```

```python
w = 20   # cámbienlo por 30 después
ver_costo(w)
```

## 4. La U

**¿Con qué w creen que el costo es más chico?** Escriban su predicción en la celda y luego corran la gráfica.

*Tip: cada punto de la curva es una recta completa; el más bajo es la que menos se equivoca.*

```python
mi_fondo = None   # su predicción, por ejemplo 26
```

```python
ws = np.linspace(15, 35, 2001)
Js = [costo(x, y, w) for w in ws]

plt.plot(ws, Js)
for w in (20, 25, 30):
    plt.scatter(w, costo(x, y, w), zorder=3)
    plt.annotate(f"J({w}) = {costo(x, y, w):.2f}", (w, costo(x, y, w)), textcoords="offset points", xytext=(6, 6))
if mi_fondo is not None:
    plt.axvline(mi_fondo, linestyle="--", color="gray", label=f"su predicción: {mi_fondo}")
    plt.legend()
plt.xlabel("w (pesos por kilo)")
plt.ylabel("J(w)")
plt.title("Cada punto de la curva es una recta completa")
plt.show()
```

Ahora que lo busque la computadora: el $w$ de la curva con el menor $J$.

```python
w_fondo = ws[np.argmin(Js)]
print(f"El fondo está en w = {w_fondo:.2f}, con J = {min(Js):.2f}")
```
